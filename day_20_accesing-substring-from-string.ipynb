```python
# concept of indexing
```

```python
s="Hello"
print(s)
```

```python
print(s[1])
```

```python
# Type of indexing
# -> Positive indexing- start from 0 for first character in string and 1 for second char and so on(left to right) 
# -> Negative indexing- start from last char for -1 index and -2 for second last char and so on(right to left)
```

```python
print(s[-1])
```

```python
print(s[-4])
```

```python
# Slicing- extract multiple character from string at a time
s="Hello World"
print(s[0:5])
```

```python
print(s[2:])
```

```python
print(s[:7])
```

```python
print(s[:])
```

```python
print(s[2:6:2])
```

```python
print(s[0:6:-1])
```

```python
print(s[-6:-1:2])
```

```python
print(s[::-1])
```
