```python
#arithmatic operator
x=5
y=2
print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x%y)
print(x**y)
print(x//y) #integer division
```

```python
#comparision operator
print(x>y)
print(x>=y)
print(x==y)
print(x!=y)
print(x<y)
print(x<=y)
```

```python
#logical operator
x=True
y=False
print(x or y)
print(x and y)
print(not x)
```

```python
#Bitwise operator
x=2
y=3
print(x & y)
print(x | y)
print(x ^ y)
print(x>>1)
print(x<<2)
print(~x)
```

```python
#Asignment operator
x=3
x+=3
print(x)
```

```python
#identity operator
a=3
b=3
print(a is b)
```

```python
a="Hello"
b="Hello"
print(a is b)
```

```python
a=[1,2,3]
b=[1,2,3]
print(a is b)
```
