```python
#find / Index
```

```python
"it is raining today".find("i") #find return the first index where the particular substring is present if not found it return -1;
```

```python
"it is raining today".endswith("day") #return true if the string is ends with substring
```

```python
"it is raining today".startswith("ing") #return true if the string is start with substring
```
