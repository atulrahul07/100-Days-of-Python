```python
import random
jackpot=random.randint(1,100)
guess=int(input("Guess number"))

counter=1;
while guess!=jackpot:
    if guess<jackpot:
        print("Guess higher number")
    else:
        print("Guess lower number")

    guess=int(input("Guess number"))
    counter+=1

print("You guess it correct in ",counter,"attempts")
```
